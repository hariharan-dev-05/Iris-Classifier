```python
x = df.drop('species', axis = 1)
y = df['species']
display(x)
display(y)
```

```python
from sklearn.discriminant_analysis import QuadraticDiscriminantAnalysis
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score

# 1. Load and split data
X_train, X_test, y_train, y_test = train_test_split(
    x, y, test_size = 0.2, random_state = 42
)

# Check the split
print("x_train:", X_train.shape)
print("x_test: ", X_test.shape)
print("y_train:", y_train.shape)
print("y_test: ", y_test.shape)

# 2. Initialize and fit the QDA model
qda = QuadraticDiscriminantAnalysis()
qda.fit(X_train, y_train)

# 3. Make predictions and evaluate
y_pred = qda.predict(X_test)
print("Accuracy:", accuracy_score(y_test, y_pred))
```

```python
# 1. Enter custom flower measurements
custom_data = pd.DataFrame({
    'sepal_length': [5.1],
    'sepal_width': [3.5],
    'petal_length': [1.4],
    'petal_width': [0.2]
})

# 2. Predict the flower species
prediction = qda.predict(custom_data)

# 3. Display the result
print("Custom Input:")
print(custom_data)

print("\nPredicted Species:", prediction[0])

# 4. Display prediction probabilities
probabilities = qda.predict_proba(custom_data)

probability_df = pd.DataFrame(
    probabilities,
    columns=qda.classes_
)

print("\nPrediction Probabilities:")
print(probability_df.round(4))
```

```python
import joblib

# Bundle everything into a single dictionary
bundle = {
    'model': qda,
    'features': list(df.columns)
}

# Save into a single pickle file
joblib.dump(bundle, 'iris.pkl')
print("Successfully saved model, and features to 'iris.pkl'!")
```
